```python
# Import Libraries

import numpy as np

import pandas as pd

import matplotlib.pyplot as plt

import seaborn as sns

import warnings

warnings.filterwarnings('ignore')
```

```python
# Loading the data set

df = pd.read_csv("/content/cars_dataset.csv")
```

```python
df['Year']=df['Year'].fillna(df['Year'].median())
```

```python
df['Colour']=df['Colour'].fillna(df['Colour'].mode())
```

```python
df.isnull().sum()
```

```python
# Univariate Analysis

sns.distplot(df['Year'])
```

```python
# Outlier treatment

Q1 = df['Year'].quantile(0.25)

Q3 = df['Year'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR

upper_bound = Q3 + 1.5 * IQR

df = df[(df['Year'] >= lower_bound) & (df['Year'] <= upper_bound)]
```

```python
# Univariate Analysis

sns.distplot(df['Year'])
```

```python
#bivariate analysis
sns.lineplot(x = 'Year', y = 'Colour', data = df)
```
